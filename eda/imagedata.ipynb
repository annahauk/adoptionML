```python

def print_tree(directory, prefix=""):
    # List files and folders, ignoring hidden files like .DS_Store
    items = sorted([i for i in os.listdir(directory) if not i.startswith('.')])
    
    for index, item in enumerate(items):
        path = os.path.join(directory, item)
        is_last = (index == len(items) - 1)
        
        # Set visual anchors
        connector = "└── " if is_last else "├── "
        print(f"{prefix}{connector}{item}")
        
        # Recursively print subdirectories
        if os.path.isdir(path):
            if item == "test_images" or item == "train_images" or item == "test_metadata" or item == "train_metadata" or item == "test_sentiment" or item == "train_sentiment":
                print(f"{prefix}{indent}└── [Contents of {item} not displayed]")
            else:
                indent = "    " if is_last else "│   "
                print_tree(path, prefix + indent)

# Run the function using your Kaggle path variable
print_tree(path)
```

```python
# populate images from train_images and test_images into a dataframe
import os


df_images = pd.DataFrame()

for root, dirs, files in os.walk(path):
    if root == path + "/train_images":
        for f in files:
            df_images = df_images.concat({"image": f, "split": "train"}, ignore_index=True)
    elif root == path + "/test_images":
        for f in files:
            df_images = df_images.concat({"image": f, "split": "test"}, ignore_index=True)
            
```
