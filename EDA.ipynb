## Episode-aware feasibility checks

The local tables support an intake-time outcome model and a length-of-stay analysis. Because an animal can appear in multiple intake episodes, rows must be paired in chronological order within `Animal ID`; a direct merge creates a many-to-many join. The model features below are restricted to information available at intake. Outcome-side fields are used only to create labels and retrospective operational metrics.

```python
# Pair each intake episode with the next unused outcome for the same animal.
# This avoids the many-to-many row inflation caused by merging on Animal ID alone.
intake = df_intakes.copy()
outcome = df_outcomes.copy()
intake["intake_dt"] = pd.to_datetime(intake["DateTime"], errors="coerce", format="mixed", utc=True)
outcome["outcome_dt"] = pd.to_datetime(outcome["DateTime"], errors="coerce", format="mixed", utc=True)

outcome_by_animal = {
    animal_id: group.sort_values("outcome_dt").to_dict("records")
    for animal_id, group in outcome.groupby("Animal ID", sort=False)
}
paired_rows = []
for animal_id, intake_group in intake.sort_values("intake_dt").groupby("Animal ID", sort=False):
    intake_records = intake_group.sort_values("intake_dt").to_dict("records")
    outcome_records = outcome_by_animal.get(animal_id, [])
    outcome_position = 0

    for intake_record in intake_records:
        while (
            outcome_position < len(outcome_records)
            and outcome_records[outcome_position]["outcome_dt"] < intake_record["intake_dt"]
        ):
            outcome_position += 1

        paired = {
            "Animal ID": animal_id,
            "intake_dt": intake_record["intake_dt"],
            "outcome_dt": pd.NaT,
            "Outcome Type": pd.NA,
        }
        for column in ["Intake Type", "Intake Condition", "Animal Type", "Sex upon Intake", "Age upon Intake", "Breed", "Color"]:
            paired[column] = intake_record[column]

        if outcome_position < len(outcome_records):
            matched_outcome = outcome_records[outcome_position]
            paired["outcome_dt"] = matched_outcome["outcome_dt"]
            paired["Outcome Type"] = matched_outcome["Outcome Type"]
            outcome_position += 1

        paired_rows.append(paired)

episodes = pd.DataFrame(paired_rows)
episodes["length_of_stay_days"] = (
    episodes["outcome_dt"] - episodes["intake_dt"]
).dt.total_seconds() / 86400

episodes["adopted"] = episodes["Outcome Type"].eq("Adoption")
known_outcomes = episodes["Outcome Type"].notna() & episodes["Outcome Type"].ne("Missing")
valid_stays = known_outcomes & episodes["length_of_stay_days"].ge(0)

print(f"Intake episodes: {len(episodes):,}")
print(f"Episodes with a known outcome: {known_outcomes.sum():,} ({known_outcomes.mean():.1%})")
print(f"Valid non-negative stays: {valid_stays.sum():,} ({valid_stays.mean():.1%})")
print(f"Unmatched or invalid stays: {(~valid_stays).sum():,}")
print("\nOutcome distribution among known outcomes:")
print(episodes.loc[known_outcomes, "Outcome Type"].value_counts(normalize=True).mul(100).round(1).to_string())
print("\nLength-of-stay quantiles for valid episodes (days):")
print(episodes.loc[valid_stays, "length_of_stay_days"].quantile([0, .25, .5, .75, .9, .95, .99]).round(1).to_string())
print("\nAdoption rate by animal type:")
print(episodes.loc[known_outcomes].groupby("Animal Type")["adopted"].agg(["mean", "count"]).assign(mean=lambda frame: frame["mean"].mul(100).round(1)).rename(columns={"mean": "adoption_rate_pct"}).sort_values("adoption_rate_pct", ascending=False).to_string())
```
